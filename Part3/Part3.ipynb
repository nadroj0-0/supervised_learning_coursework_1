### 9. *Kernel modification* Consider the function $K_c(x,z) := c + \sum_{i=1}^n x_i z_i$ where $x, z \in \mathbb{R}^n$.
#### a) For what values of $c \in \mathbb{R}$ is $K_c$ a positive semidefinite kernel? Give an argument supporting your claim.

If $c = 0$, then $K_c(x,z) = \sum_{i=1}^n x_i z_i = <x, z>$. Therefore, $K_c$ is a positive semidefinite kernel.

If $c > 0$, then $K_c(x,z) = c + \sum_{i=1}^n x_i z_i = <\phi(x), \phi(z)>$ with $\phi(x) = (\sqrt{c}, x_1, \ldots, x_n)$. Therefore, $K_c$ is a positive semidefinite kernel.

If $c < 0$, the previous argument cannot be used, as $sqrt(c) \notin \mathbb{R}$ when $c < 0$. In this case, it can be seen that, when taking $x=0$, $K_c(x, x) = K_c(0, 0) = c < 0$. Therefore, $K_c$ is not a positive semidefinite kernel when $c < 0$.

#### b) Suppose we use $K_c$ as a kernel function with linear regression (least squares). Explain how $c$ influences the solution.

When $c=0$, the corresponding linear regression is purely linear and forced through the origin, as there is no actual feature map applied over the data, while te case $c=1$ is equal to the standard linear regression with bias.

Therefore, the variable $c$ just adds a bias to the linear regression. The value of $c$ only causes the weight associated to the bias to be divided by $\sqrt{c}$ with respect to the standard model with a bias. This causes high weights in the bias to be less pensalised as $c$ increases when regularising linear regression.

### 10. Suppose we perform linear regression with a Gaussian kernel $K_{\beta}(x,t) = e^{-\beta \left\lVert x-t \right\rVert^2}$ to train a classifier on a dataset $(x_1, y_1), \ldots, (x_n, y_n) \in \mathbb{R}^n \times {-1, 1}$. Thus obtaining a function $f: \mathbb{R}^n \rightarrow \mathbb{R}$ which is of the form $f(t) = \sum_{i=1}^m \alpha_i K_{\beta}(x_i, t)$. This classifier depends on the parameter $\beta$ selected for the kernel. In what scenario will the chose $\beta$ enable the trained linear classifier to simulate a 1-Nearest Neighbor classifier trained on the same dataset?

With the Gaussian kernel, the classifier is the following: $f_{\beta}(t) = \sum_{i=1}^m \alpha_i e^{- \beta \left\lVert x_i-t \right\rVert^2}$. 

Intuitively, when $\beta \rightarrow \infty$, the term $e^{- \beta \left\lVert x_i-t \right\rVert^2}$ will tend to $0$, except for the cases in which $\left\lVert x_i-t \right\rVert^2 \approx 0$. Therefore, only the points which are a very small distance away from the point $t$ will be considered. For a big enough value of $\beta$, the nearest neighbor will have a much higher value for $e^{- \beta \left\lVert x_i-t \right\rVert^2}$ than the rest, enabling the trained linear classifier to simulate 1-NN.

Let's give a more rigorous proof. Let $x_n$ be the nearest neighbor, so that $\left\lVert x_n-t \right\rVert^2 \leq \left\lVert x_i-t \right\rVert^2$ for all $i \neq n$. 

With this notation, $\lim_{\beta \rightarrow \infty} \frac{K_{\beta}(x_i, t)}{K_{\beta}(x_n, t)} = \lim_{\beta \rightarrow \infty} \frac{e^{- \beta \left\lVert x_i-t \right\rVert^2}}{e^{- \beta \left\lVert x_n-t \right\rVert^2}} = \lim_{\beta \rightarrow \infty} e^{-\beta(\left\lVert x_i-t \right\rVert^2 - \left\lVert x_n-t \right\rVert^2)} = 0$ for $i \neq n$, as $\left\lVert x_i-t \right\rVert - \left\lVert x_n-t \right\rVert > 0$.

Therefore, $f_{\beta}(t) = \alpha_n K_{\beta}(x_n, t) + \sum_{i\neq n} \alpha_i K_{\beta}(x_i, t) = K_{\beta}(x_n, t)[\alpha_n + \sum_{i\neq n} \alpha_i \frac{K_{\beta}(x_i, t)}{K_{\beta}(x_n, t)}]$. Taking limits: $$\lim_{\beta \rightarrow \infty} f_{\beta}(t) = \lim_{\beta \rightarrow \infty} K_{\beta}(x_n, t)[\alpha_n + \sum_{i\neq n} \alpha_i \frac{K_{\beta}(x_i, t)}{K_{\beta}(x_n, t)}] = \alpha_n \lim_{\beta \rightarrow \infty} K_{\beta}(x_n, t).$$

We have proved that $\lim_{\beta \rightarrow \infty} f_{\beta}(t)$ only depends on the nearest neighbor. Now we must prove that the value assigned to the point t is exactly the label of the nearest neighbor.

If we take a fixed, but very high value of $\beta$, $f_{\beta}(t) \approx \alpha_n K_{\beta}(x_n, t)$, as we have proved beforehand. Training the linear regressor is equivalent to minimizing the least-squares error, taking into account that the nearest neighbor to a training point is itself. Therefore, the following least-squares error formula is obtained: $\sum_{i=1}^m (y_i - f_{\beta}(x_i) )^2 = \sum_{i=1}^m (y_i - \alpha_i K_{\beta}(x_i, x_i))^2 = \sum_{i=1}^m (y_i - \alpha_i)^2$. Thus, $\alpha_i = y_i$ for all $i$, so $f_{\beta}(t) \approx \alpha_n K_{\beta}(x_n, t) \approx y_n$ as $K_{\beta}(x_n, t) \approx 1$ if $x_n$ and $t$ are close enough.
